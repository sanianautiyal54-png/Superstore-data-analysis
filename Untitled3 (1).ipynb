```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

df = pd.read_excel(
    "Superstore_Excel_Cleaned.xlsx",
    sheet_name="Cleaned_Data"
)

df.head()
```

```python
# Check missing values
df.isnull().sum()
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()

print("Total Sales:", total_sales)
print("Total Profit:", total_profit)
```

```python
category_sales = (
    df.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print(category_sales)
```

```python
plt.figure(figsize=(8,5))
category_sales.plot(kind="bar")
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
category_profit = (
    df.groupby("Category")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

print(category_profit)
```

```python
plt.figure(figsize=(8,5))
category_profit.plot(kind="bar")
plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
region_sales = (
    df.groupby("Region")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print(region_sales)
```

```python
plt.figure(figsize=(8,5))
region_sales.plot(kind="bar")
plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])

monthly_sales = (
    df.groupby(df["Order Date"].dt.to_period("M"))["Sales"]
    .sum()
)

monthly_sales.index = monthly_sales.index.astype(str)

print(monthly_sales)
```

```python
plt.figure(figsize=(12,5))
plt.plot(monthly_sales.index, monthly_sales.values)
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
top_customers = (
    df.groupby(["Customer ID", "Customer Name"])["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(top_customers)
```

```python
plt.figure(figsize=(10,6))
top_customers.sort_values().plot(kind="barh")
plt.title("Top 10 Customers by Sales")
plt.xlabel("Sales")
plt.ylabel("Customer")
plt.tight_layout()
plt.show()
```

```python
top_products = (
    df.groupby("Product Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(top_products)
```

```python
plt.figure(figsize=(10,6))
top_products.sort_values().plot(kind="barh")
plt.title("Top 10 Products by Sales")
plt.xlabel("Sales")
plt.ylabel("Product")
plt.tight_layout()
plt.show()
```

```python
subcategory_profit = (
    df.groupby("Sub-Category")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

print(subcategory_profit)
```

```python
plt.figure(figsize=(10,6))
subcategory_profit.plot(kind="bar")
plt.title("Profit by Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Profit")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Discount"], df["Profit"], alpha=0.5)
plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")
plt.tight_layout()
plt.show()
```

```python
# Sales outliers using IQR
Q1 = df["Sales"].quantile(0.25)
Q3 = df["Sales"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

sales_outliers = df[
    (df["Sales"] < lower_limit) |
    (df["Sales"] > upper_limit)
]

print("Number of Sales Outliers:", len(sales_outliers))
```

```python
# Profit outliers using IQR
Q1_profit = df["Profit"].quantile(0.25)
Q3_profit = df["Profit"].quantile(0.75)

IQR_profit = Q3_profit - Q1_profit

lower_profit = Q1_profit - 1.5 * IQR_profit
upper_profit = Q3_profit + 1.5 * IQR_profit

profit_outliers = df[
    (df["Profit"] < lower_profit) |
    (df["Profit"] > upper_profit)
]

print("Number of Profit Outliers:", len(profit_outliers))
```

```python
# Correlation between numeric variables
correlation = df[["Sales", "Quantity", "Discount", "Profit"]].corr()

print(correlation)
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(correlation, annot=True)
plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()
```

```python
# Key business metrics

total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Order ID"].nunique()
total_quantity = df["Quantity"].sum()

average_order_value = total_sales / total_orders
profit_margin = (total_profit / total_sales) * 100

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Orders:", total_orders)
print("Total Quantity:", total_quantity)
print("Average Order Value:", round(average_order_value, 2))
print("Profit Margin:", round(profit_margin, 2), "%")
```

```python
# Highest-sales category

category_sales = df.groupby("Category")["Sales"].sum()

highest_category = category_sales.idxmax()
highest_category_sales = category_sales.max()

print("Highest Sales Category:", highest_category)
print("Sales:", round(highest_category_sales, 2))
```

```python
# Highest-profit category

category_profit = df.groupby("Category")["Profit"].sum()

highest_profit_category = category_profit.idxmax()
highest_category_profit = category_profit.max()

print("Highest Profit Category:", highest_profit_category)
print("Profit:", round(highest_category_profit, 2))
```

```python
# Highest-sales region

region_sales = df.groupby("Region")["Sales"].sum()

highest_region = region_sales.idxmax()
highest_region_sales = region_sales.max()

print("Highest Sales Region:", highest_region)
print("Sales:", round(highest_region_sales, 2))
```

```python
# Most profitable sub-category

subcategory_profit = df.groupby("Sub-Category")["Profit"].sum()

best_subcategory = subcategory_profit.idxmax()
best_subcategory_profit = subcategory_profit.max()

print("Most Profitable Sub-Category:", best_subcategory)
print("Profit:", round(best_subcategory_profit, 2))
```
